### **Problemática**:
El sector financiero ha experimentado una transformación significativa con el avance de las tecnologías digitales, particularmente en la forma en que se otorgan y gestionan los préstamos. Esta evolución ha permitido a las instituciones financieras ofrecer créditos de manera más eficiente, fomentando la inversión y facilitando el acceso a capital para personas y empresas. Sin embargo, este crecimiento también ha traído consigo nuevos desafíos, especialmente relacionados con la gestión del riesgo crediticio.

Uno de los principales problemas que enfrentan los bancos y entidades financieras es el incumplimiento en el pago de los préstamos por parte de los clientes. Cuando un prestatario no cumple con sus obligaciones de pago, la institución incurre en pérdidas económicas significativas. A gran escala, estas pérdidas no solo afectan la estabilidad financiera de las entidades prestamistas, sino que también impactan negativamente el sistema económico en general, ralentizando el crecimiento económico del país y generando desconfianza en el mercado financiero.

Por ello, es clave contar con herramientas predictivas que permitan detectar tempranamente patrones anómalos o conductas de riesgo en solicitantes de crédito. Este proyecto plantea una primera aproximación al uso de técnicas de machine learning como herramienta para identificar patrones de comportamiento anómalos en solicitantes de crédito, permitiendo anticipar posibles incumplimientos de pago y apoyar la toma de decisiones en la gestión del riesgo financiero. La solución busca no solo reducir las pérdidas por impagos, sino también aportar a un sistema financiero más eficiente, justo y sostenible.
